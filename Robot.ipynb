```python
import random
import time

estado_robot = {}

def reiniciar_estado():
    """Reinicia el estado simulado del robot (posición, tiempo, batería)."""
    global estado_robot
    estado_robot = {
        "distancia_a_casa": 0,   # metros
        "tiempo_transcurrido": 0,  # minutos
        "bateria": 100,          # %
        "modo_regreso": False,
    }

# =====================================================
# PASO 1: CONFIGURACIÓN DEL ROBOT — funciones auxiliares
# =====================================================

def consultar_mapa_del_barrio():
    print("🗺️  Consultando el mapa del barrio...")

def elegir_ruta_segura(evitar_calles_con_mucho_trafico=True):
    ruta = ["casa", "parque", "calle_tranquila", "casa"]
    print(f"✅ Ruta segura elegida: {' -> '.join(ruta)}")
    return ruta

def definir_punto_partida_y_regreso(casa):
    print(f"📍 Punto de partida y de regreso definido: {casa}")

def definir_horario_del_paseo():
    while True:
        try:
            tiempo = int(input("⏱️  [INTERACTIVO] Ingrese el tiempo máximo del paseo (en min): "))
            if tiempo > 0:
                print(f"⏱️  Tiempo máximo de paseo configurado: {tiempo} min")
                return tiempo
            print("❌ Error: El tiempo debe ser mayor a 0.")
        except ValueError:
            print("❌ Error: Ingrese un valor numérico válido.")

def definir_nivel_minimo_bateria():
    while True:
        try:
            nivel = int(input("🔋 [INTERACTIVO] Ingrese el % mínimo de batería seguro para retornar: "))
            if 0 < nivel <= 100:
                print(f"🔋 Nivel mínimo seguro de batería configurado: {nivel}%")
                return nivel
            print("❌ Error: Ingrese un porcentaje entre 1 y 100.")
        except ValueError:
            print("❌ Error: Ingrese un valor numérico válido.")


# =====================================================
# PASO 2: VERIFICACIÓN DE HERRAMIENTAS E INSUMOS — funciones auxiliares
# (Incluye botella de agua según reglamento de Madrid)
# =====================================================

def tiene_correa():
    return True

def tiene_bolsas():
    return True

def tiene_botella_agua():
    return True

def mostrar_mensaje(mensaje):
    print(f"💬 {mensaje}")

def alertar_falta_material():
    print("🚨 Alerta: falta material para iniciar el paseo.")


# =====================================================
# PASO 3: EJECUCIÓN DE LAS TAREAS — funciones auxiliares
# =====================================================

# --- Condiciones climáticas y activación ---

def esta_lloviendo():
    return random.random() < 0.1  # 10% de probabilidad, para poder ver el caso

def posponer_paseo():
    print("⏸️  Paseo pospuesto hasta que mejore el tiempo.")

def se_activa_por_boton():
    return True

def se_activa_por_app():
    return False

def esperar_activacion():
    print("⏳ Esperando activación por botón o por app...")


# --- Avance y posición ---

def avanzar_por(ruta):
    global estado_robot
    if estado_robot["modo_regreso"]:
        estado_robot["distancia_a_casa"] = max(0, estado_robot["distancia_a_casa"] - 40)
        print(f"🚶 Volviendo hacia casa... distancia restante: {estado_robot['distancia_a_casa']} m")
    else:
        estado_robot["distancia_a_casa"] += 30
        print(f"🚶 Avanzando por la ruta... distancia a casa: {estado_robot['distancia_a_casa']} m")
    estado_robot["tiempo_transcurrido"] += 1
    estado_robot["bateria"] = max(0, estado_robot["bateria"] - random.randint(2, 5))

def obtener_posicion_actual():
    return "casa" if estado_robot["distancia_a_casa"] == 0 else "paseo"


# --- Evitar charcos / superficies irregulares ---

def detectar_obstaculo():
    return random.choice(["charco", "ninguno", "ninguno", "ninguno"])  # ~25% charco

def detectar_superficie_irregular():
    return random.random() < 0.1

def recalcular_ruta(evitar_obstaculo=True):
    if random.random() < 0.8:  # 80% de las veces se encuentra una ruta alternativa
        nueva_ruta = ["posicion_actual", "calle_alternativa", "casa"]
        print(f"🔄 Ruta alternativa encontrada: {' -> '.join(nueva_ruta)}")
        return nueva_ruta
    print("❌ No se encontró ninguna ruta alternativa.")
    return None

def detener_robot():
    print("🛑 Robot detenido.")

def alertar_obstaculo_insalvable():
    print("🚨 Alerta: obstáculo insalvable, no hay ruta disponible.")

def redirigir_a_casa():
    global estado_robot
    estado_robot["modo_regreso"] = True
    print("↩️  Redirigiendo el robot hacia casa.")
    return ["posicion_actual", "casa"]


# --- Recoger bolsitas ---

def detectar_necesidades_perro():
    return random.random() < 0.3

def posicionar_brazo_robotico():
    print("🦾 Posicionando el brazo robótico...")

def coger_bolsita():
    print("🛍️  Bolsita cogida del dispensador.")

def recoger_residuos():
    print("🧹 Residuos recogidos.")

def buscar_papelera_cercana():
    if random.random() < 0.85:
        papelera = "papelera de la calle Mayor"
        print(f"🗑️  Papelera encontrada: {papelera}")
        return papelera
    print("❌ No se encontró ninguna papelera cercana.")
    return None

def depositar_bolsita(papelera):
    print(f"✅ Bolsita depositada en: {papelera}")

def guardar_bolsita_temporalmente():
    print("🎒 Bolsita guardada temporalmente en el compartimento del robot.")

def reanudar_marcha():
    print("▶️  Reanudando la marcha.")


# --- Limpiar orina ---

def detectar_orina_perro():
    return random.random() < 0.3

def rociar_agua_botella():
    print("💧 Rociando agua de la botella para limpiar.")


# --- Control de tiempo y batería ---

def consultar_tiempo_restante(tiempo_maximo):
    return tiempo_maximo - estado_robot["tiempo_transcurrido"]

def calcular_tiempo_para_volver(posicion_actual):
    return estado_robot["distancia_a_casa"] // 30 + 1

def consultar_nivel_bateria():
    return estado_robot["bateria"]

def ajustar_velocidad(aumentar=True):
    print("⚡ Ajustando velocidad" + (" (aumentando)" if aumentar else ""))

def confirmar_llegada_a_casa():
    print("🏠 ¡El robot ha llegado a casa!")

def detener_proceso():
    print("⛔ Proceso detenido: no se puede iniciar el paseo.")


# =====================================================
# Programa principal
# =====================================================

def crear_robot_paseador_perros():
    reiniciar_estado()

    # =====================================================
    # PASO 1: CONFIGURACIÓN DEL ROBOT
    # =====================================================

    consultar_mapa_del_barrio()
    ruta = elegir_ruta_segura(evitar_calles_con_mucho_trafico=True)
    definir_punto_partida_y_regreso("casa")

    # --- Asignar horario e inputs (AHORA MANUALES) ---
    tiempo_maximo = definir_horario_del_paseo()
    nivel_minimo_seguro = definir_nivel_minimo_bateria()

    # =====================================================
    # PASO 2: VERIFICACIÓN DE HERRAMIENTAS E INSUMOS
    # =====================================================

    if tiene_correa() and tiene_bolsas() and tiene_botella_agua():
        material_completo = True
        mostrar_mensaje("Material completo. Robot listo para activarse.")
    else:
        alertar_falta_material()
        material_completo = False

    # =====================================================
    # PASO 3: EJECUCIÓN DE LAS TAREAS
    # =====================================================

    if material_completo:

        if esta_lloviendo():
            mostrar_mensaje("Está lloviendo. Paseo pospuesto.")
            posponer_paseo()
        else:
            if se_activa_por_boton() or se_activa_por_app():

                paseo_activo = True
                posicion_actual = "casa"
                ha_salido_de_casa = False

                MAX_ITERACIONES_SIMULACION = 200
                contador_iteraciones = 0

                while paseo_activo:
                    # ANIMACIÓN: Pausa de 1.5 segundos entre cada vuelta de la manzana
                    time.sleep(1.5)

                    contador_iteraciones += 1
                    if contador_iteraciones > MAX_ITERACIONES_SIMULACION:
                        print("⚠️  Límite de iteraciones de la simulación alcanzado.")
                        break

                    print(f"\n--- Vuelta {contador_iteraciones} del paseo ---")

                    if detectar_obstaculo() == "charco" or detectar_superficie_irregular():
                        ruta_alternativa = recalcular_ruta(evitar_obstaculo=True)
                        if ruta_alternativa:
                            ruta = ruta_alternativa
                        else:
                            detener_robot()
                            alertar_obstaculo_insalvable()
                            ruta = redirigir_a_casa()

                    avanzar_por(ruta)
                    posicion_actual = obtener_posicion_actual()

                    if posicion_actual != "casa":
                        ha_salido_de_casa = True

                    if detectar_necesidades_perro():
                        detener_robot()
                        posicionar_brazo_robotico()
                        coger_bolsita()
                        recoger_residuos()
                        papelera = buscar_papelera_cercana()
                        if papelera:
                            depositar_bolsita(papelera)
                        else:
                            guardar_bolsita_temporalmente()
                        reanudar_marcha()

                    if detectar_orina_perro():
                        rociar_agua_botella()

                    tiempo_restante = consultar_tiempo_restante(tiempo_maximo)
                    tiempo_para_volver = calcular_tiempo_para_volver(posicion_actual)
                    nivel_bateria = consultar_nivel_bateria()

                    if tiempo_restante <= tiempo_para_volver or nivel_bateria <= nivel_minimo_seguro:
                        ajustar_velocidad(aumentar=True)
                        ruta = redirigir_a_casa()

                    if ha_salido_de_casa and posicion_actual == "casa":
                        confirmar_llegada_a_casa()
                        paseo_activo = False

            else:
                esperar_activacion()

    else:
        mostrar_mensaje("No se puede iniciar el paseo. Falta material.")
        detener_proceso()

# ¡Arranca la magia, batería!
crear_robot_paseador_perros()
```
